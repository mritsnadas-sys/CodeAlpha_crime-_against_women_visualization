```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset (Replace 'crime_data.csv' with your actual file)
df = pd.read_csv(r"C:\Users\SWAGATIKA\Documents\crime against women in India in last 5 years.csv")


# 1. Scatter Plot for Data Distribution
plt.figure(figsize=(8,5))
sns.scatterplot(x=df.columns[0], y=df.columns[1], data=df)
plt.title("Scatter Plot")
plt.show()

# 2. Line Plot to Show Trends
plt.figure(figsize=(8,5))
sns.lineplot(x=df.columns[0], y=df.columns[1], data=df)
plt.title("Line Plot - Trends")
plt.show()

# 3. Histogram for Distribution Analysis
plt.figure(figsize=(8,5))
sns.histplot(df[df.columns[1]], bins=10, kde=True)
plt.title("Histogram")
plt.show()

# 4. Box Plot for Outlier Detection
plt.figure(figsize=(8,5))
sns.boxplot(y=df[df.columns[1]])
plt.title("Box Plot - Outlier Detection")
plt.show()

# 5. Violin Plot for Distribution and Density
plt.figure(figsize=(8,5))
sns.violinplot(y=df[df.columns[1]])
plt.title("Violin Plot - Density")
plt.show()

# 6. Heatmap for Correlation Matrix
plt.figure(figsize=(8,5))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap="coolwarm")
plt.title("Heatmap - Correlation Matrix")
plt.show()

# 7. Pair Plot to Show Relationships
sns.pairplot(df)
plt.show()

# 8. Bar Plot for Categorical Comparisons
plt.figure(figsize=(8,5))
sns.barplot(x=df.columns[0], y=df.columns[1], data=df)
plt.xticks(rotation=90)
plt.title("Bar Plot - Categorical Comparison")
plt.show()

# 9. Swarm Plot for Categorical Scatter
plt.figure(figsize=(8,5))
sns.swarmplot(x=df.columns[0], y=df.columns[1], data=df)
plt.xticks(rotation=90)
plt.title("Swarm Plot - Categorical Scatter")
plt.show()
```
